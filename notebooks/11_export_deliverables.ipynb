## Imports

````python
import json
import shutil

from pathlib import Path
from datetime import datetime
````

## Define directories

````python
PROJECT_DIR = Path(
    "/content/tkh-hierarchy-project"
)


EXPORT_DIR = (
    PROJECT_DIR
    /
    "deliverables"
)


SUBDIRS = [

    "data",

    "hierarchy",

    "evaluation",

    "reports",

    "metadata"

]


for folder in SUBDIRS:

    (
        EXPORT_DIR
        /
        folder
    ).mkdir(
        parents=True,
        exist_ok=True
    )


print(
    EXPORT_DIR
)
````

## Check available artifacts

````python
for path in PROJECT_DIR.rglob("*"):

    if path.is_file():

        print(path)
````

## Copy hierarchy outputs

````python
source = (
    PROJECT_DIR
    /
    "artifacts"
    /
    "hierarchy"
)


destination = (
    EXPORT_DIR
    /
    "hierarchy"
)


if source.exists():

    shutil.copytree(

        source,

        destination,

        dirs_exist_ok=True

    )


print(
    "Hierarchy exported."
)
````

## Copy evaluation outputs

````python
evaluation_files = [

    "t1_statistics.json",

    "retrieval_evaluation_results.csv"

]


for filename in evaluation_files:


    matches = list(
        PROJECT_DIR.rglob(filename)
    )


    if matches:

        shutil.copy(

            matches[0],

            EXPORT_DIR
            /
            "evaluation"
            /
            filename

        )


        print(
            "Exported:",
            filename
        )

    else:

        print(
            "Not found:",
            filename
        )
````

## Export summary metadata

````python
metadata = {

    "project":

        "Multi-Resolution Semantic Abstraction over an Evolving Knowledge Hypergraph",


    "generated":

        datetime.now().isoformat(),


    "pipeline":

        [

            "01_load_and_validate_data",

            "02_temporal_snapshot_construction",

            "03_descriptive_statistics",

            "04_semantic_representation",

            "05_hypergraph_objective",

            "06_multiresolution_coarsening",

            "07_temporal_coupling",

            "08_labels",

            "09_intrinsic_evaluation",

            "10_extrinsic_retrieval"

        ]

}


with open(

    EXPORT_DIR
    /
    "metadata"
    /
    "project_metadata.json",

    "w",

    encoding="utf-8"

) as f:


    json.dump(

        metadata,

        f,

        indent=2

    )


print(
    "Metadata saved."
)
````

## Generate manifest

````python
manifest = []


for file in EXPORT_DIR.rglob("*"):

    if file.is_file():

        manifest.append({

            "file":

                str(
                    file.relative_to(
                        EXPORT_DIR
                    )
                ),

            "size_bytes":

                file.stat().st_size

        })


with open(

    EXPORT_DIR
    /
    "metadata"
    /
    "manifest.json",

    "w",

    encoding="utf-8"

) as f:


    json.dump(

        manifest,

        f,

        indent=2

    )


print(
    "Files exported:",
    len(manifest)
)
````
